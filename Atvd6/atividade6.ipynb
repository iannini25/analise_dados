```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
path1 = "Tabela5-sem_emprego_2012.csv"
path2 = "Tabela5-sem_emprego_2026.csv"
path3 = "Tabela 1.1.1.xls"
```

```python
t2012 = pd.read_csv(path1, sep=";", encoding="utf-8-sig", decimal=",")
t2012 = t2012.rename(columns={"Desocupados - homens (2012 T1)": "homens_2012", "Desocupados - mulheres (2012 T1)": "mulheres_2012"})
t2012
```

```python
t2026 = pd.read_csv(path2, sep=";", encoding="utf-8-sig", decimal=",")
t2026 = t2026.rename(columns={"Desocupados - homens (2026 T1)": "homens_2026", "Desocupados - mulheres (2026 T1)": "mulheres_2026"})
t2026
```

```python
comp = t2012.merge(t2026, on=["Sigla", "Código", "Estado"], how="inner")
comp
```

```python
bruto = pd.read_excel(path3, engine="xlrd", header=None)
horas = bruto.iloc[8:41].copy()
horas.columns = [
    "uf_regiao",
    "total",
    "total_branca",
    "total_preta_parda",
    "homem_branca",
    "homem_preta_parda",
    "mulher_branca",
    "mulher_preta_parda",
]
horas = horas.reset_index(drop=True)
for col in horas.columns[1:]:
    horas[col] = pd.to_numeric(horas[col], errors="coerce")
horas
```

```python
regioes = ["Norte", "Nordeste", "Sudeste", "Sul", "Centro-Oeste"]
horas = horas[~horas["uf_regiao"].isin(["Brasil"] + regioes)]
horas = horas.rename(columns={"uf_regiao": "Estado"})
horas
```

```python
comp = comp.merge(horas, on="Estado", how="inner")
comp
```

```python
ordem = comp.sort_values("mulheres_2026")["Estado"]

longo = comp.melt(
    id_vars=["Estado", "Sigla"],
    value_vars=["mulheres_2012", "mulheres_2026"],
    var_name="Ano",
    value_name="Participacao_mulheres",
)
longo["Ano"] = longo["Ano"].map({"mulheres_2012": "2012 T1", "mulheres_2026": "2026 T1"})
longo.head()
```

```python
fig, ax = plt.subplots(figsize=(10, 10))
sns.barplot(
    data=longo,
    y="Estado",
    x="Participacao_mulheres",
    hue="Ano",
    order=ordem,
    ax=ax,
)
ax.axvline(50, color="red", linestyle="--", linewidth=1)
ax.set_xlabel("Participação das mulheres entre as pessoas desocupadas (%)")
ax.set_ylabel("")
ax.set_title("Desocupação: participação feminina em 2012 T1 e 2026 T1")
ax.legend(title="Ano")
fig.tight_layout()
plt.show()
```
